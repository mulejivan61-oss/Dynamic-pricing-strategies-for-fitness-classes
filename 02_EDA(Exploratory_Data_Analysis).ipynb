Time Analysis 


```python
#create features
data["Month"]= data["BookingEndDate"].dt.month_name()
data["Day"]=data["BookingEndDate"].dt.day_name()
#time slot
data["Hour"]=data["BookingStartTime"].dt.hour
def time_slot(hour):
    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    else:
        return "Evening"
    
data["TimeSlot"] = data["Hour"].apply(time_slot)
```

```python
data["TimeSlot"].value_counts()
```

Feature Engineering

```python
#Occupancy Rate
data["Occupancy Rate"] = (data["Number Booked"] / data["MaxBookees"]) * 100

data[["Number Booked", "MaxBookees", "Occupancy Rate"]].head()
```

```python
#Occupancy Analysis
plt.figure(figsize=(8,2))
sns.boxplot(x=data["Occupancy Rate"])
plt.title("Occupancy Rate Boxplot")
plt.show()
```

```python
#revenue
data["Revenue"] = data["Price (INR)"] * data["Number Booked"]
```

```python
#revenue by timeSlot
order = ["Morning", "Afternoon", "Evening"]

revenue_slot = (
    data.groupby("TimeSlot")["Revenue"]
      .sum()
      .reindex(order)
      .reset_index()
)

plt.figure(figsize=(8,5))
sns.barplot(data=revenue_slot, x="TimeSlot", y="Revenue")

plt.title("Total Revenue by Time Slot")
plt.xlabel("Time Slot")
plt.ylabel("Total Revenue (INR)")

plt.show()
```

```python
#revenue by activity
top_revenue = (
    data.groupby("ActivityDescription")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
      .reset_index()
)

plt.figure(figsize=(12,6))
sns.barplot(
    data=top_revenue,
    x="Revenue",
    y="ActivityDescription"
)

plt.title("Top 10 Activities by Revenue")
plt.xlabel("Total Revenue (INR)")
plt.ylabel("Activity")

plt.show()
```

```python
# Revenue by Location

revenue_location = (
    data.groupby("ActivitySiteID")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)

plt.figure(figsize=(10,5))

sns.barplot(
    data=revenue_location,
    x="ActivitySiteID",
    y="Revenue"
)

plt.title("Revenue by Location")
plt.xlabel("Activity Site ID")
plt.ylabel("Total Revenue (INR)")
plt.xticks(rotation=45)

plt.show()
```

```python
#Correlation analysis
plt.figure(figsize=(9,7))
sns.heatmap(data.corr(numeric_only=True), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix")
plt.show()
```
