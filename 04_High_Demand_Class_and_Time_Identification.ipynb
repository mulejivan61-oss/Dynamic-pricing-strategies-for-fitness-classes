Review of EDA Findings

The Exploratory Data Analysis (EDA) provided valuable insights into customer booking behavior, class popularity, and revenue generation.

Key Findings
Morning time slots recorded the highest number of customer bookings.
Wednesday showed the highest booking demand, whereas Sunday recorded comparatively lower demand.
Body Conditioning, Aquafit, Body Balance, and Body Combat were among the most popular fitness classes.
The HXP location generated the highest revenue.
Pilates and Body Combat contributed significantly to overall revenue.
A positive correlation was observed between the number of bookings and revenue, indicating that increased customer participation directly improves business revenue.

Review of Price Elasticity Results

The Price Elasticity Model was developed using Linear Regression to estimate customer bookings based on various pricing and operational factors.

Model Performance
R² Score: 0.7858
RMSE: 4.39
Findings
The model explained approximately 78.6% of the variation in customer bookings.
Pricing, class type, booking time, and location significantly influenced customer demand.
The model demonstrated good predictive performance and can support pricing-related decisions.

Review of Demand Forecasting Results

An ARIMA model was developed to forecast future customer bookings.

Model Performance
MAE: 74.37
MSE: 13781.93
RMSE: 117.40
Findings
The forecasting model successfully captured the overall booking demand pattern.
The forecast closely followed the actual booking trend for most testing dates.
The model can be used to estimate future demand and support proactive business planning.

Key Factors Influencing Customer Demand

Based on the combined findings from EDA, Price Elasticity Analysis, and Demand Forecasting, the following factors were identified as major contributors to customer demand.

# Factor	                                     # Impact on Demand
Morning Time Slot	                             High Demand

Body Conditioning & Aquafit	                     High Customer Preference
HXP Location	                                 High Revenue Generation
Booking Volume	                                 Strong Positive Impact on Revenue
Historical Booking Trend	                     Useful for Demand Forecasting

Business Recommendations:

Introduce dynamic pricing during peak booking periods to maximize revenue.
